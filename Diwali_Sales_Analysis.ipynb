# Exploratory Data Analysis

```python
df.columns
```

### Gender

```python
# plotting a bar chart for Gender and it's count

sns.countplot(x='Gender', data= df)
```

```python
ax = sns.countplot(x='Gender', data=df)

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# plotting a bar chart for gender vs total amount

sales_gen = df.groupby(['Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending = False)
sns.barplot(x='Gender', y='Amount', data= sales_gen)
```

From the above analysis, we can observe that the number of female buyers(7832) is significantly higher than male buyers(3407)during the Diwali sale. Moreover, the total purchase amount by females is also considerably higher than males, indicating that woman not only shop more in numbers but also spend more overall during the festive season.

### Age

```python
ax = sns.countplot(data = df, x= 'Age Group', hue = 'Gender')
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
# Total Amount vs Age Group
sales_age = df.groupby(['Age Group'], as_index= False)['Amount'].sum().sort_values(by='Amount', ascending=False)
sns.barplot(x='Age Group', y = 'Amount', data= sales_age)
```

From the above analysis, we can see that the age group of 26-35 years has the highest number of buyers, with females (3269)significantly outnumbering males(1272) in this group. Additionally, the 26-35 age group also contributes the highest total purchase amount compared to all other age groups, followed by 36-45 and 18-25. This suggests that young to mid-aged working women are most active and highest spending shoppers during the Diwali festive season.

### State

```python
df.columns
```

```python
# total number of orders from top 10 states

sales_state = df.groupby(['State'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10
                                                                                                            )

sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data = sales_state, x = 'State',y= 'Orders')
```

```python
# total amount/sales from top 10 states

sales_state = df.groupby(['State'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

sns.set(rc={'figure.figsize':(15,5)})
sns.barplot(data = sales_state, x = 'State',y= 'Amount')Df
```

From the above analysis, we can clearly observe that Uttar Pradesh leads significantly in both the total number of orders and total sales amount among all Indian states, followed by Maharashtra and Karnataka in second and third positions respectively. This indicates the northern and western regions of India are the biggest contributers to Diwali festive sales, with Uttar Pradesh being the dominant market by a considerable margin compared to other states.


### Marital Status

```python
ax = sns.countplot(data = df, x = 'Marital_Status')

sns.set(rc={'figure.figsize':(7,5)})
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_state = df.groupby(['Marital_Status', 'Gender'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.set(rc={'figure.figsize':(6,5)})
sns.barplot(data = sales_state, x = 'Marital_Status',y= 'Amount', hue='Gender')
```

The analysis indicates that married women represent the largest buying segment, with a count of 6,518, and demonstrates significantly higher purchasing power compared to other groups. Therefore, marries woaman should be considered the primary target demographic for maximising Diwali sales revenue.

### Occupation

```python
sns.set(rc={'figure.figsize':(20,5)})
ax = sns.countplot(data = df, x = 'Occupation')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_state = df.groupby(['Occupation'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = sales_state, x = 'Occupation',y= 'Amount')
```

The analysis reveals that buyers employes in the IT Sector, Healthcare, and Aviation industries have the highest purchase count as well as the highest total spending amount, making them the most commercially significant occupational segments during Diwali sales.

### Product Category

```python
sns.set(rc={'figure.figsize':(20,5)})
ax = sns.countplot(data = df, x = 'Product_Category')

for bars in ax.containers:
    ax.bar_label(bars)
```

```python
sales_state = df.groupby(['Product_Category'], as_index=False)['Amount'].sum().sort_values(by='Amount', ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = sales_state, x = 'Product_Category',y= 'Amount')
```

The analysis indicates that food, clothing & Apparel, and Electronics & Gadgets are the top performing product categories, both in terms of number of orders and total revenue generated. These three categories collectively dominate Diwali sales and should be prioritized in inventory planning and marketing strategies.

```python
sales_state = df.groupby(['Product_ID'], as_index=False)['Orders'].sum().sort_values(by='Orders', ascending=False).head(10)

sns.set(rc={'figure.figsize':(20,5)})
sns.barplot(data = sales_state, x = 'Product_ID',y= 'Orders')
```
